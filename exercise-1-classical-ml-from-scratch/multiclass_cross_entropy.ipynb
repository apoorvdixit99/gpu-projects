```python
import numpy as np

def softmax(x):
    # Prevent overflow
    x = x - np.max(x)
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)
```

```python
import numpy as np

def cross_entropy(logits, y):
    # logits: (n_samples, n_classes)
    # y: (n_samples,)

    # Numerically stable softmax
    probabilities = np.array([
        softmax(row) for row in logits
    ])

    # Extract probability of correct class
    n = len(y)

    # correct_probs = probabilities[np.arange(n), y]
    correct_probs = []
    for i in range(n):
        correct_probs.append(probabilities[i, y[i]])

    # Calculate average loss
    loss = -np.mean(np.log(correct_probs))

    return loss
```

```python
# Tests for multiclass cross-entropy implemented in the notebook

def softmax_rowwise(logits):
    shift = logits - np.max(logits, axis=1, keepdims=True)
    exp = np.exp(shift)
    return exp / exp.sum(axis=1, keepdims=True)

def expected_cross_entropy(logits, y):
    probs = softmax_rowwise(logits)
    n = len(y)
    correct = probs[np.arange(n), y]
    return -np.mean(np.log(correct))

# Test cases
tests = [
    (np.array([[2.0, 1.0, 0.1],
               [0.5, 2.5, 0.3]]), np.array([0, 1])),
    (np.array([[1000.0, 1000.0, 999.0],
               [1200.0, 1199.0, 1180.0]]), np.array([1, 0])),
]

for i, (logits, y) in enumerate(tests, 1):
    loss = cross_entropy(logits, y)
    expected = expected_cross_entropy(logits, y)
    try:
        np.testing.assert_allclose(loss, expected, rtol=1e-6, atol=1e-8)
        print(f"Test {i} passed: loss={loss:.8f}")
    except AssertionError:
        print(f"Test {i} FAILED")
        print("computed loss:", loss)
        print("expected   :", expected)
        raise
```

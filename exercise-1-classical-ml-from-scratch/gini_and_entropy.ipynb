```python
def entropy(y):
    counts = {}
    for label in y:
        counts[label] = counts.get(label, 0) + 1
    probabilities = [ count/len(y) for count in counts.values() ]
    return -sum(p * np.log(p) for p in probabilities)
```

```python
def gini(y):
    counts = {}
    for label in y:
        counts[label] = counts.get(label, 0) + 1
    probabilities = [ count/len(y) for count in counts.values() ]
    return 1-sum(p**2 for p in probabilities)
```

```python
def information_gain(parent, left, right):
    n = len(parent)
    parent_entropy = entropy(parent)
    weighted_entropy = (
        len(left) / n * entropy(left)
        + len(right) / n * entropy(right)
    )
    return parent_entropy - weighted_entropy
```

```python
import numpy as np

# Test cases for entropy()
tests = {
    (1, 1, 1, 1): 0.0,
    (0, 1): np.log(2),
    (0, 0, 1, 1): np.log(2),
    ("a", "a", "b"): -(2/3) * np.log(2/3) - (1/3) * np.log(1/3),
}

for values, expected in tests.items():
    result = entropy(list(values))
    print(f"values={values}, entropy={result}, expected={expected}")
    assert np.isclose(result, expected)

print("All entropy tests passed.")
```

```python
# Test cases for gini()
gini_tests = {
    (1, 1, 1, 1): 0.0,
    (0, 1): 0.5,
    (0, 0, 1, 1): 0.5,
    ("a", "a", "b"): 4 / 9,
}

for values, expected in gini_tests.items():
    result = gini(list(values))
    print(f"values={values}, gini={result}, expected={expected}")
    assert np.isclose(result, expected)

print("All gini tests passed.")
```

```python
import numpy as np

def softmax(x):
    # Prevent overflow
    x = x - np.max(x)
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)
```

```python
def log_softmax(x):
    # return x - np.log(np.sum(np.exp(x)))
    max_x = np.max(x)
    log_sum_exp = max_x + np.log(
        np.sum(np.exp(x - max_x))
    )
    return x - log_sum_exp
```

```python
# Sample input data
x = np.array([1.0, 2.0, 3.0])

# Test softmax
probs = softmax(x)
print("softmax(x):", probs)

# Test log_softmax
log_probs = log_softmax(x)
print("log_softmax(x):", log_probs)

# Check mathematical relationship: log_softmax(x) = log(softmax(x))
assert np.allclose(log_probs, np.log(probs))
assert np.isclose(np.sum(probs), 1.0)
assert np.all(probs > 0)
```

```python
# Demonstrate cases where the naive softmax/log-softmax formulas overflow/underflow
# while the numerically stable versions work correctly.

def naive_softmax(x):
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)

def naive_log_softmax(x):
    return x - np.log(np.sum(np.exp(x)))

# These inputs trigger overflow/underflow in the naive formulas
bad_inputs = [
    np.array([1000.0, 1001.0, 1002.0]),    # huge positive values -> overflow
    np.array([-1000.0, -1001.0, -1002.0]), # huge negative values -> underflow
    np.array([1000.0, 1000.0, 1000.0]),    # all same large value -> exp overflows
]

for x in bad_inputs:
    print(f"Input x = {x}")
    print("Naive softmax:", naive_softmax(x))
    print("Stable softmax:", softmax(x))
    print("Naive log-softmax:", naive_log_softmax(x))
    print("Stable log-softmax:", log_softmax(x))
    print("Stable softmax sum:", np.sum(softmax(x)))
    print("Finite stable softmax:", np.all(np.isfinite(softmax(x))))
    print("-" * 60)

# A few strict checks
x = np.array([1000.0, 1001.0, 1002.0])
naive_probs = naive_softmax(x)
stable_probs = softmax(x)

assert not np.all(np.isfinite(naive_probs)), "Naive softmax should fail on large positive values"
assert np.all(np.isfinite(stable_probs)), "Stable softmax should remain finite"
assert np.allclose(stable_probs.sum(), 1.0), "Stable softmax probabilities should sum to 1"
assert np.allclose(log_softmax(x), np.log(stable_probs)), "Log-softmax should match log(stable softmax)"
```
